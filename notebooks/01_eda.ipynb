```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

pd.set_option("display.max_columns", None)
```

```python
# Load datasets
train = pd.read_csv("../data/train.csv")
calendar = pd.read_csv("../data/calendar_events.csv")
submission_template = pd.read_csv("../data/forecast_submission.csv")

print("train shape:", train.shape)
print("calendar shape:", calendar.shape)
print("submission template shape:", submission_template.shape)

train.head()
```

```python
train.groupby("store_id")["date"].agg(["min", "max", "nunique"])
```

```python
store0 = train[train["store_id"] == 0].sort_values("date")

plt.figure(figsize=(12, 4))
plt.plot(store0["date"], store0["revenue"])
plt.title("Store 0 — Daily Revenue")
plt.xlabel("Date")
plt.ylabel("Revenue")
plt.show()

store1 = train[train["store_id"] == 1].sort_values("date")

plt.figure(figsize=(12, 4))
plt.plot(store1["date"], store1["revenue"])
plt.title("Store 1 — Daily Revenue")
plt.xlabel("Date")
plt.ylabel("Revenue")
plt.show()

store2 = train[train["store_id"] == 2].sort_values("date")

plt.figure(figsize=(12, 4))
plt.plot(store2["date"], store2["revenue"])
plt.title("Store 2 — Daily Revenue")
plt.xlabel("Date")
plt.ylabel("Revenue")
plt.show()

store3 = train[train["store_id"] == 3].sort_values("date")

plt.figure(figsize=(12, 4))
plt.plot(store3["date"], store3["revenue"])
plt.title("Store 3 — Daily Revenue")
plt.xlabel("Date")
plt.ylabel("Revenue")
plt.show()
```
